# C. CATEGORY-AVERAGE VISUALIZATION

```python
CatAve = pd.read_excel('board2.xlsx')
```

```python
CatAve['Average'] = CatAve[['Math','GEAS','Electronics', 'Communication']].mean(axis=1)
```

```python
AveTrack = CatAve.groupby('Track')['Average'].mean()
AveTrack
```

```python
AveGender = CatAve.groupby('Gender')['Average'].mean()
AveGender
```

```python
AveHometown = CatAve.groupby('Hometown')['Average'].mean()
AveHometown
```

```python
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(16,6))
axes[0].bar(AveTrack.index, AveTrack.values)
axes[1].bar(AveGender.index, AveGender.values)
axes[2].bar(AveHometown.index, AveHometown.values)
axes[0].set(title='AveTrack', ylabel='Mean',
xlabel='Categories')
axes[1].set(title='AveGender', ylabel='Mean',
xlabel='Categories')
axes[2].set(title='AveHometown', ylabel='Mean',
xlabel='Categories')
```

### In the AveTrack, the category with the highest sample mean is Communication.

### In the AveGender, the Male category has the highest mean.

### In the AveHometown, Luzon has the highest sample mean.
